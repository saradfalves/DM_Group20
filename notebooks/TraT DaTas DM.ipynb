### Quantas inconsistências existem, uma vez corrigido o 01-01-1970?

```python
ref = pd.Timestamp("2024-01-01")

idx_1970 = df_nodups.index[df_nodups["membership_end"] == pd.Timestamp("1970-01-01")]
df_nodups["end_was_1970"] = df_nodups.index.isin(idx_1970)

df_nodups["membership_end"] = df_nodups["membership_end"].replace(pd.Timestamp("1970-01-01"), ref)

# Entre as três variáveis
idx_start_after_end = df_nodups.index[
    df_nodups["membership_start"] > df_nodups["membership_end"]]

idx_start_after_last_swipe = df_nodups.index[
    df_nodups["membership_start"] > df_nodups["last_badge_swipe"]]

idx_end_before_last_swipe = df_nodups.index[
    df_nodups["membership_end"] < df_nodups["last_badge_swipe"]]

print("INCONSISTÊNCIAS ENTRE VARIÁVEIS")
print("membership_start > membership_end:", idx_start_after_end.size)
print("membership_start > last_badge_swipe:", idx_start_after_last_swipe.size)
print("membership_end < last_badge_swipe:", idx_end_before_last_swipe.size)

print("------------------------------------------------------------------")

# Entre as variáveis e a data de referência
idx_start_after_ref = df_nodups.index[
    df_nodups["membership_start"] > ref]

idx_end_after_ref = df_nodups.index[
    df_nodups["membership_end"] > ref]

idx_last_swipe_after_ref = df_nodups.index[
    df_nodups["last_badge_swipe"] > ref]

print("INCONSISTÊNCIAS FACE À DATA DE REFERÊNCIA")
print("membership_start > ref:", idx_start_after_ref.size)
print("membership_end > ref:", idx_end_after_ref.size)
print("last_badge_swipe > ref:", idx_last_swipe_after_ref.size)
```

```python
# "last_badge_swipe > ref: 33" -> podemos tratar e normalizar o last_badge_swipe para a data de referência - evidenciar que ainda estão ativos
df_nodups.loc[idx_last_swipe_after_ref, "last_badge_swipe"] = ref
```

```python
# agora vamos resolver a questão do start e end:

df_nodups.loc[(df_nodups["membership_start"] > df_nodups["membership_end"]), ["membership_start", "membership_end", "last_badge_swipe", 
                                                                              "days_since_last_visit",  "lifetime_value", "n_renewals", "total_visits", "avg_weekly_visits"]]

# apresentar total_visits e avg_weekly_visits permite-nos estimar o número de semanas que a membership durou e,
# admitindo, por hipótese, que last_badge_swipe não contém erros, uma vez corrigidos as datas posteriores a ref, podemos usá-lo para comparar
# se a estimated_date for muito anterior ao last_badge_swipe, podeomos ter um problema; se for relativamente próxima (ou posterior), podemos usá-la e corrigir.
# naturalmente, também podemos esta variável para comparar com a membership_start, q tb registou 93 problemas de incompatibilidade.

# A ordem correta parece-me ser:
# 1. Corrigir os last_badge_swipe;
# 2. Corrigir os days_since_last_visit e, com a relação, corrigir os membership_end;
# 3. Finalmente, a relação entre membership_end e estimated_weeks, corrigir membership_start.
```

```python
df_nodups["days_since_last_visit"].describe()
```

```python
# Comparação das hipóteses para days_since_last_visit

# Hipótese 1: membership_end substituída pela data de referência
df_nodups["days_since_ref"] = (
    ref - df_nodups["last_badge_swipe"]
).dt.days

# Hipótese 2: membership_end como data de referência
df_nodups["days_since_end"] = (
    df_nodups["membership_end"] - df_nodups["last_badge_swipe"]
).dt.days

# Diferenças absolutas
df_nodups["diff_ref"] = (
    df_nodups["days_since_last_visit"] - df_nodups["days_since_ref"]
).abs()

df_nodups["diff_end"] = (
    df_nodups["days_since_last_visit"] - df_nodups["days_since_end"]
).abs()

# Resumo comparativo
for hypothesis, diff in [
    ("DATA DE REFERÊNCIA", "diff_ref"),
    ("FIM DA SUBSCRIÇÃO", "diff_end")
]:
    valid = df_nodups[diff].dropna()

    print(f"\nHIPÓTESE: {hypothesis}")
    print(f"Observações válidas: {valid.size}")
    print(f"Correspondências exatas: {(valid == 0).sum()}")
    print(f"Correspondências (%): {(valid == 0).mean() * 100:.2f}%")
    print(f"Diferença média: {valid.mean():.2f} dias")
    print(f"Diferença mediana: {valid.median():.2f} dias")
```

Assumimos, pelos resultados obtidos acima, que `days_since_last_visit` representa o número de dias decorridos entre a última visita registada e o fim da subscrição.

$$\text{days\_since\_last\_visit} = \text{membership\_end} - \text{last\_badge\_swipe} $$

Consequentemente, podemos também obter:

$$ \text{membership\_end} = \text{last\_badge\_swipe} + \text{days\_since\_last\_visit} $$

```python
# Existem claramente dois tipos de problemas ao analisar os valores: em alguns casos, o problema está na membership_end (que antecede o membership_start)
# e noutros o problema está no last_badge_swipe (pq somando estimated_weeks ao membership_start bate mais ou menos com o membership_end)
# Para corrigir, vamos separar:
# basicamente, no grupo B como o last_badge_swipe vai para lá do membership_end, é razoável assumir que usaram o plano até ao fim, ou seja, days_since_last_visit = 0
# (tb são só 25 linhas, não é nada de outro mundo)

grupo_a = (df_nodups["days_since_last_visit"] < 0) & (df_nodups["membership_start"] > df_nodups["membership_end"])

df_nodups.loc[grupo_a, ["membership_start", "membership_end", "last_badge_swipe",
                         "days_since_last_visit", "estimated_weeks", "n_renewals", "lifetime_value"]]
```

```python
df_nodups.loc[grupo_a, "membership_end"] = (
    df_nodups.loc[grupo_a, "membership_start"]
    + pd.to_timedelta(df_nodups.loc[grupo_a, "estimated_weeks"] * 7, unit="D").dt.round("D")
)
```

```python
# 1) fim nunca depois da data de referência (só grupo A)
fim_depois_ref = grupo_a & (df_nodups["membership_end"] > ref)
df_nodups.loc[fim_depois_ref, "membership_end"] = ref

# 2) fim nunca antes da última visita (só grupo A)
fim_antes_visita = grupo_a & (df_nodups["membership_end"] < df_nodups["last_badge_swipe"])
df_nodups.loc[fim_antes_visita, "membership_end"] = df_nodups.loc[fim_antes_visita, "last_badge_swipe"]

# 3) recalcular os dias (só grupo A)
df_nodups.loc[grupo_a, "days_since_last_visit"] = (
    df_nodups.loc[grupo_a, "membership_end"] - df_nodups.loc[grupo_a, "last_badge_swipe"]
).dt.days
```

```python
# 1) identificar e guardar os índices ANTES de alterar
grupo_b = (df_nodups["days_since_last_visit"] < 0) & (df_nodups["membership_end"] >= df_nodups["membership_start"])
idx_grupo_b = df_nodups.index[grupo_b]

len(idx_grupo_b)   # 25

# 2) a última visita passa a ser o fim
df_nodups.loc[idx_grupo_b, "last_badge_swipe"] = df_nodups.loc[idx_grupo_b, "membership_end"]

# 3) recalcular os dias com a fórmula
df_nodups.loc[idx_grupo_b, "days_since_last_visit"] = (
    df_nodups.loc[idx_grupo_b, "membership_end"] - df_nodups.loc[idx_grupo_b, "last_badge_swipe"]
).dt.days
```

```python
df_nodups.loc[df_nodups["membership_start"] > df_nodups["membership_end"], 
        ["membership_start", "membership_end", "last_badge_swipe",
         "days_since_last_visit", "estimated_weeks", "n_renewals", "lifetime_value"]]

# para os casos que em que days_since_last_visit não é 999 (os casos que tinham ficado por resolver anteriormente), podemos recuperar facilmente
# qual será o valor de membership_end, usando, novamente, estimated_weeks.

filtro = (df_nodups["membership_start"] > df_nodups["membership_end"]) & (df_nodups["days_since_last_visit"] < 999)
idx_filtro = df_nodups.index[filtro]

df_nodups.loc[idx_filtro, "membership_end"] = df_nodups.loc[idx_filtro, "last_badge_swipe"] + pd.to_timedelta(df_nodups.loc[idx_filtro, "days_since_last_visit"], unit="D")

# limitar à data de referência
depois_ref = df_nodups.index.isin(idx_filtro) & (df_nodups["membership_end"] > ref)
df_nodups.loc[depois_ref, "membership_end"] = ref

# nessas linhas o fim mudou, por isso os dias têm de ser recalculados
df_nodups.loc[depois_ref, "days_since_last_visit"] = (
    df_nodups.loc[depois_ref, "membership_end"] - df_nodups.loc[depois_ref, "last_badge_swipe"]
).dt.days
```

```python
# É notório que todas as membership_end correspondem praticamente à estimated_end_date, pelo que a variável que deverá ser corrigida é last_badge_swipe
# através da relação last_badge_swipe = membership_end - days_since_last_visit:

filtro_swipe = (df_nodups["membership_end"] < df_nodups["last_badge_swipe"]) & (df_nodups["membership_start"] < df_nodups["membership_end"])
idx_swipe = df_nodups.index[filtro_swipe]

df_nodups.loc[idx_swipe, "last_badge_swipe"] = df_nodups.loc[idx_swipe, "membership_end"] - pd.to_timedelta(df_nodups.loc[idx_swipe, "days_since_last_visit"], unit="D")     
```

```python
df_nodups.loc[(df_nodups["membership_start"] > df_nodups["last_badge_swipe"]) & (df_nodups["days_since_last_visit"] < 999), 
        ["membership_start", "membership_end", "estimated_end_date", "last_badge_swipe",
         "days_since_last_visit", "total_visits", "estimated_weeks", "n_renewals", "lifetime_value"]]

# mais uma vez, o problema parece ser last_badge_swipe! note-se que, por ex, na linha 778, o valor 496 (days) é superior à duração da subscrição.

df_nodups["tenure_days"] = (df_nodups["membership_end"] - df_nodups["membership_start"]).dt.days

filtro_inicio = (
    (df_nodups["membership_start"] > df_nodups["last_badge_swipe"])
    & (df_nodups["days_since_last_visit"] < 999)
    & (df_nodups["tenure_days"] >= df_nodups["days_since_last_visit"])
)

idx_inicio = df_nodups.index[filtro_inicio]

df_nodups.loc[idx_inicio, "last_badge_swipe"] = df_nodups.loc[idx_inicio, "membership_end"] - pd.to_timedelta(df_nodups.loc[idx_inicio, "days_since_last_visit"], unit="D")
```

```python

# 1) guardar os índices ANTES de alterar
filtro_999 = (df_nodups["membership_start"] > df_nodups["membership_end"]) & (df_nodups["days_since_last_visit"] == 999)
idx_999_fim = df_nodups.index[filtro_999]

len(idx_999_fim)

# 2) fim = fim estimado, e no máximo a data de referência
df_nodups.loc[idx_999_fim, "membership_end"] = df_nodups.loc[idx_999_fim, "estimated_end_date"]
df_nodups.loc[idx_999_fim, "membership_end"] = df_nodups.loc[idx_999_fim, "membership_end"].where(df_nodups.loc[idx_999_fim, "membership_end"] <= ref, ref)

# 3) recalcular os dias, que deixam de ser 999
df_nodups.loc[idx_999_fim, "days_since_last_visit"] = (
    df_nodups.loc[idx_999_fim, "membership_end"] - df_nodups.loc[idx_999_fim, "last_badge_swipe"]).dt.days

df_nodups.loc[idx_999_fim, ["membership_start", "membership_end", "last_badge_swipe", "days_since_last_visit"]]
```

```python
sem_recuperacao = (
    (df_nodups["last_badge_swipe"] < df_nodups["membership_start"])
    & (df_nodups["days_since_last_visit"] > df_nodups["tenure_days"])
)

idx_sem_recuperacao = df_nodups.index[sem_recuperacao]

# df_nodups.loc[idx_sem_recuperacao, "last_badge_swipe"] = pd.NaT
# df_nodups.loc[idx_sem_recuperacao, "days_since_last_visit"] = float("nan")

df_nodups.loc[
    idx_sem_recuperacao,
    ["membership_start", "membership_end", "estimated_end_date", "last_badge_swipe",
     "tenure_days", "days_since_last_visit", "total_visits", "estimated_weeks"]
]
```

```python
# 1) grupos de frequência
grupo_freq = pd.cut(
    df_nodups["avg_weekly_visits"], bins=[-0.01, 0.5, 1, 2, 3, 100], labels=["<0.5", "0.5-1", "1-2", "2-3", ">3"])

# 2) mediana dos dias em cada grupo (as linhas em falta são ignoradas)
mediana_grupo = df_nodups.groupby(grupo_freq, observed=True)["days_since_last_visit"].transform("median")

# 3) preencher os dias nas linhas sem recuperação
df_nodups.loc[idx_sem_recuperacao, "days_since_last_visit"] = mediana_grupo[idx_sem_recuperacao]

# 4) os dias nunca podem passar a duração da subscrição
passa_tenure = df_nodups.index.isin(idx_sem_recuperacao) & (df_nodups["days_since_last_visit"] > df_nodups["tenure_days"])
df_nodups.loc[passa_tenure, "days_since_last_visit"] = df_nodups.loc[passa_tenure, "tenure_days"]

# 5) reconstruir o last_badge_swipe
df_nodups.loc[idx_sem_recuperacao, "last_badge_swipe"] = df_nodups.loc[idx_sem_recuperacao, "membership_end"] - pd.to_timedelta(df_nodups.loc[idx_sem_recuperacao, "days_since_last_visit"], unit="D")
```
